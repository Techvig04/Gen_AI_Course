```python
import google.generativeai as genai
from google.colab import userdata

GOOGLE_API_KEY=userdata.get('Gemini_API_Key')
genai.configure(api_key=GOOGLE_API_KEY)
```

```python
gemini_model = genai.GenerativeModel('gemini-3.5-flash')
prompt_text = "I want to know the 7 wonder of the world"

response = gemini_model.generate_content(prompt_text)
print(response.text)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", google_api_key=GOOGLE_API_KEY)
```

```python
from langchain_core.output_parsers import StrOutputParser
restaurant_name_prompt = PromptTemplate(
    input_variables=["cuisine"],
    template="I want to open an {cuisine} restaurant. Suggest five creative names for it."
)

name_chain = restaurant_name_prompt | llm | StrOutputParser()

print("Suggested Indian Restaurant Names:")
response_name = name_chain.invoke({"cuisine": "Indian"})
print(response_name)
```

```python
import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field

os.environ["G_A_K"]=userdata.get("Gemini_API_Key")
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash",google_api_key=os.environ["G_A_K"],temperature=0.7)
class ResturantPlan(BaseModel):
  resturant_names:list[str]= Field(description="5 Creative Resturant Names ")
  tagline:str =Field(description ="A catchy tagline for the resturant")
  popular_dishes:list[str]=Field(description="3 Signature menu Items")
parser = JsonOutputParser(pydantic_object=ResturantPlan)
prompt=PromptTemplate(
    template=""" You are a resturant consultant .Create a concept for {cuisine}
    resturant located in {location} with a {theme} theme.
    {format_instructions}""",
    input_variables=["cuisine","location","theme"],
    partial_variables={"format_instructions":parser.get_format_instructions()},

)
chain = prompt | llm | parser
response = chain.invoke({"cuisine":"Indian",
                         "location":"New Delhi",
                         "theme":"traditional"

})
print("---Suggested Names---")
for name in response["resturant_names"]:
  print(f"-{name}")
print("\n---tagline---")
print(response["tagline"])
print("\n --- Signature Dishes ---")
for dish in response["popular_dishes"]:
  print(f"-{dish}")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

chatbot_llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", google_api_key=os.environ["G_A_K"], temperature=0.7)
```

```python
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful AI assistant. Answer all questions concisely."),
        ("human", "{question}"),
    ]
)

chatbot_chain = chat_template | chatbot_llm | StrOutputParser()

user_question = "What is the capital of India?"
response = chatbot_chain.invoke({"question": user_question})
print(f"User: {user_question}")
print(f"AI Assistant: {response}")

user_question_2 = "Tell me a fun fact about Python programming."
response_2 = chatbot_chain.invoke({"question": user_question_2})
print(f"\nUser: {user_question_2}")
print(f"AI Assistant: {response_2}")
```
